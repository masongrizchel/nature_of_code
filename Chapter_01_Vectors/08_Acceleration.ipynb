## 1.8 Vector Motion: Acceleration
Acceleration is the rate of change of velocity. It affects velocity, which in turn affects location, creating a trickle-down effect. We can use a constant acceleration or a random acceleration each cycle.

```python
# Example 1.8 & 1.9: Motion 101 with Acceleration
class MoverAccel:
    def __init__(self):
        self.location = PVector(width/2, height/2)
        self.velocity = PVector(0, 0)
        self.topspeed = 10

    def update(self):
        # Random acceleration:
        self.acceleration = PVector.random2D() 
        self.acceleration.mult(random(2))
        
        self.velocity.add(self.acceleration)
        self.velocity.limit(self.topspeed) # Constrains the magnitude
        self.location.add(self.velocity)
```
