## 1.4 More Vector Math & 1.5 Vector Magnitude
Multiplication scales a vector (changing its length without affecting direction) by multiplying each component by a scalar. The magnitude (or length) of a vector is calculated using the Pythagorean theorem.

```python
# Example 1.4 & 1.5: Vector multiplication and magnitude
def setup():
    size(200, 200)
    smooth()

def draw():
    background(255)
    mouse = PVector(mouseX, mouseY)
    center = PVector(width/2, height/2)
    mouse.sub(center)
    
    # Multiplication
    mouse.mult(0.5) 
    
    # Magnitude
    m = mouse.mag()
    fill(0)
    rect(0, 0, m, 10)
    
    translate(width/2, height/2)
    line(0, 0, mouse.x, mouse.y)
```
