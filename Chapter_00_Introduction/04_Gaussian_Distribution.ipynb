# 1.4 & 1.5 Gaussian and Custom Distributions
This notebook covers normal (Gaussian) distributions and custom distributions (like the Monte Carlo method) to generate random numbers with specific probabilities[cite: 11, 15, 16].

```python
import random
import matplotlib.pyplot as plt

# Example 1.4: Gaussian Distribution
# Using a mean of 320 and standard deviation of 60
gaussian_results = [random.gauss(320, 60) for _ in range(10000)]

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.hist(gaussian_results, bins=50, color='lightgray', edgecolor='black')
plt.title("Gaussian Distribution")

# Example 1.5: Custom Distribution (Monte Carlo)
def montecarlo():
    while True: # Do this until we find a qualifying random value
        r1 = random.random()     # Pick a random value
        probability = r1         # Assign a probability
        r2 = random.random()     # Pick a second random value
        
        if r2 < probability:     # Does it qualify?
            return r1

montecarlo_results = [montecarlo() for _ in range(10000)]

plt.subplot(1, 2, 2)
plt.hist(montecarlo_results, bins=50, color='gray', edgecolor='black')
plt.title("Monte Carlo Distribution")

plt.tight_layout()
plt.show()
```
