# 1.3 Probability and Non-Uniform Distributions
This notebook demonstrates a random walker with custom probabilities. Instead of a uniform 25% chance for each direction, this walker has a 40% chance of moving to the right.

```python
import matplotlib.pyplot as plt
import random

class RightWalker:
    def __init__(self):
        self.x, self.y = 0, 0
        self.history_x, self.history_y = [0], [0]

    def step(self):
        r = random.random() # A random floating point value between 0 and 1
        
        if r < 0.4:        # 40% chance of moving to the right
            self.x += 1
        elif r < 0.6:
            self.x -= 1
        elif r < 0.8:
            self.y += 1
        else:
            self.y -= 1
            
        self.history_x.append(self.x)
        self.history_y.append(self.y)

w = RightWalker()
for _ in range(10000):
    w.step()

plt.plot(w.history_x, w.history_y, color='black', linewidth=0.5)
plt.title("Walker That Tends to Move Right")
plt.show()
```
