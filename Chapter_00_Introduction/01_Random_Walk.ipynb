# 1.1 Random Walks
This notebook simulates a traditional random walk where a Walker object takes steps in four possible directions (up, down, left, right) based on random choices.

```python
import matplotlib.pyplot as plt
import random

class Walker:
    def __init__(self):
        # Initialize the Walker's starting location
        self.x = 0
        self.y = 0
        self.history_x = [self.x]
        self.history_y = [self.y]

    def step(self):
        # Pick a random number between 0 and 3
        choice = random.randint(0, 3) 
        
        if choice == 0:
            self.x += 1
        elif choice == 1:
            self.x -= 1
        elif choice == 2:
            self.y += 1
        else:
            self.y -= 1
            
        self.history_x.append(self.x)
        self.history_y.append(self.y)

w = Walker()
for _ in range(10000): # Simulating the continuous draw() loop
    w.step()

plt.figure(figsize=(6, 6))
plt.plot(w.history_x, w.history_y, color='black', linewidth=0.5)
plt.title("Traditional Random Walk")
plt.axis('equal')
plt.show()
```
