# 1.6 Perlin Noise
This notebook demonstrates a random walker using Perlin noise to generate smooth, naturally ordered sequences of pseudo-random numbers, resulting in more organic movements.

```python
import matplotlib.pyplot as plt
from noise import pnoise1

class NoiseWalker:
    def __init__(self):
        self.tx = 0.0
        self.ty = 10000.0 # Start y at 10,000 so x and y act independently
        self.history_x = []
        self.history_y = []

    def step(self):
        # Generate noise and map it (pnoise1 returns -1 to 1; map it to 0-640 for width)
        # Using a width of 640 and height of 360
        x = (pnoise1(self.tx) + 1) / 2 * 640 
        y = (pnoise1(self.ty) + 1) / 2 * 360 
        
        self.history_x.append(x)
        self.history_y.append(y)
        
        self.tx += 0.01 # Move forward through "time"
        self.ty += 0.01

w = NoiseWalker()
for _ in range(1000):
    w.step()

plt.plot(w.history_x, w.history_y, color='black', alpha=0.5)
plt.scatter(w.history_x[-1], w.history_y[-1], color='black', s=100) # Draw current position
plt.title("Perlin Noise Walker")
plt.xlim(0, 640)
plt.ylim(0, 360)
plt.gca().invert_yaxis()
plt.show()
```
