### Attractor Class and Implementation
Defines a single `Attractor` object that pulls multiple `Mover` objects towards it.

```python
class Attractor:
    def __init__(self):
        self.location = Vector(width/2, height/2)
        self.mass = 20
        self.G = 8.4 # Gravitational constant

    def attract(self, m):
        force = self.location - m.location # Vector pointing to attractor
        distance = force.magnitude
        distance = constrain(distance, 5.0, 25.0) # Constrain to avoid extremes
        
        force.normalize()
        # Gravity formula
        strength = (self.G * self.mass * m.mass) / (distance * distance)
        force *= strength
        return force

    def display(self):
        stroke(0)
        fill(175, 200)
        circle(self.location.x, self.location.y, self.mass * 2)

attractor = Attractor()

def draw():
    background(255)
    attractor.display()
    
    for m in movers:
        force = attractor.attract(m) # Calculate attraction force
        m.apply_force(force)
        m.update()
        m.display()
```
