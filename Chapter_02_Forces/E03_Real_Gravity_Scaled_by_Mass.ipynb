### Updated Draw Loop
Applies wind and a mass-scaled gravity force to all movers.

```python
def draw():
    background(255)
    wind = Vector(0.001, 0)
    
    for m in movers:
        # Scale gravity by mass to be more accurate
        gravity = Vector(0, 0.1 * m.mass) 
        
        m.apply_force(wind)
        m.apply_force(gravity)
        m.update()
        m.display()
        m.check_edges()
```
