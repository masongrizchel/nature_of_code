### Mover Class
Defines the primary object that responds to forces. Includes Newton's Second Law implementation (`apply_force`) with mass scalar.

```python
from p5 import *

class Mover:
    def __init__(self, m, x, y):
        self.mass = m
        self.location = Vector(x, y)
        self.velocity = Vector(0, 0)
        self.acceleration = Vector(0, 0)
        
    def apply_force(self, force):
        # Make a copy of the force vector before dividing by mass
        f = force.copy()
        f /= self.mass
        self.acceleration += f # Force accumulation
        
    def update(self):
        self.velocity += self.acceleration
        self.location += self.velocity
        self.acceleration *= 0 # Clear acceleration each frame
        
    def display(self):
        stroke(0)
        fill(175)
        circle(self.location.x, self.location.y, self.mass * 16)
        
    def check_edges(self):
        if self.location.x > width:
            self.location.x = width
            self.velocity.x *= -1
        elif self.location.x < 0:
            self.location.x = 0
            self.velocity.x *= -1
            
        if self.location.y > height:
            self.location.y = height
            self.velocity.y *= -1
```
