### Mover Additions & Mutual Attraction Draw Function
Add the `attract` method to the `Mover` class and define the `draw` function for mutual attraction.

```python
# Add this method to the Mover class
def attract(self, m):
    force = self.location - m.location
    distance = constrain(force.magnitude, 5.0, 25.0)
    force.normalize()
    G = 8.4
    strength = (G * self.mass * m.mass) / (distance * distance)
    force *= strength
    return force

def draw():
    background(255)
    for i, m1 in enumerate(movers):
        for j, m2 in enumerate(movers):
            if i != j: # Don't attract yourself
                force = m2.attract(m1)
                m1.apply_force(force)
        
        m1.update()
        m1.display()
```
