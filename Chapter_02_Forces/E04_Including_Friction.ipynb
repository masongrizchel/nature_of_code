### Main Draw Loop
Applies environmental forces (friction, wind, and gravity) and updates all movers.

```python
def draw():
    background(255)
    wind = Vector(0.001, 0)
    gravity = Vector(0, 0.1)
    
    for m in movers:
        c = 0.01 # Friction coefficient
        friction = m.velocity.copy()
        friction *= -1 # Opposite direction of velocity
        friction.normalize()
        friction *= c # Apply magnitude
        
        m.apply_force(friction) # Apply friction force
        m.apply_force(wind)
        m.apply_force(gravity)
        m.update()
        m.display()
        m.check_edges()
```
