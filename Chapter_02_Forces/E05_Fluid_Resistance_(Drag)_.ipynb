### Liquid Class
Defines a liquid environment that applies drag.

```python
class Liquid:
    def __init__(self, x, y, w, h, c):
        self.x, self.y, self.w, self.h = x, y, w, h
        self.c = c # Coefficient of drag
        
    def display(self):
        no_stroke()
        fill(175)
        rect(self.x, self.y, self.w, self.h)
```

Add these methods to your existing `Mover` class.

```python
class Mover:
    # ... (your existing Mover __init__ and methods here) ...
    
    def is_inside(self, l):
        # Determines if the location is inside the Liquid rectangle
        return (self.location.x > l.x and self.location.x < l.x + l.w and \
                self.location.y > l.y and self.location.y < l.y + l.h)
                
    def drag(self, l):
        speed = self.velocity.magnitude
        drag_magnitude = l.c * speed * speed # Force magnitude: Cd * v^2
        
        drag_force = self.velocity.copy()
        drag_force *= -1
        drag_force.normalize() # Force direction: -1 * velocity unit vector
        drag_force *= drag_magnitude
        self.apply_force(drag_force)
```
